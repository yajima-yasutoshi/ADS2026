#Google Colab の概要

##Google Colaboratory とは

Google Colaboratory（Google Colab）は、ブラウザ上で Jupyter Notebook を利用できるサービスである。
手元のパソコンに Python をインストールしなくても、
Pythonコードの作成と実行をすることができる。

### 主な特徴

- Python のコードをブラウザ上で実行できる
- 説明文、数式、表、プログラム、実行結果を1つのノートブックにまとめられる
- Google Drive への保存や、他の利用者との共有が容易
- CPU に加え、利用状況に応じて GPU・TPU を利用できる
- 無料で利用できるが、計算資源や連続実行時間は保証されず、制限は変動する
- 機械学習やデータ分析、教育用の演習に適している

### CPU と GPU

- **CPU（Central Processing Unit）**：幅広い処理を実行する汎用的な演算装置
- **GPU（Graphics Processing Unit）**：同じ種類の計算を多数並列に行うことが得意な演算装置

> GPU を選択しただけで処理が自動的に高速化されるわけではない。GPU 対応のライブラリを用いて
コーディングを行う必要がある。

> GPU を用いたライブラリーにはどのようなものがあるか調べてみよ。

## 注意点
- Colab の実行環境（ランタイム）は一時的な仮想マシンである。
- ファイルなどの保存を`/content` に行うことが可能である。しかし、
一定時間操作しない場合や、セッションが終了した場合、`/content` に置いたファイルや変数は消える。
- ノートブック本体は Google Drive に保存される。
- 追加インストールしたライブラリは、次回接続時に復元されない。
- パスワード、API キー、個人情報などをコードセルへ直接書かないこと。Colabのシークレット機能を用いること。左端の「鍵」のアイコンをクリックするとメニューが表示される。

#Python の基本

##変数と代入

変数は、値に名前を付けて記憶するための仕組みである。`=` は「等しい」ではなく、右側の値を左側の変数へ代入する記号である。

### 変数名の基本

- 英字、数字、アンダースコア `_` を使うのが一般的
- 数字から始めることはできない
- 大文字と小文字は区別される
- Python では日本語の変数名も使用できるが、講義では読みやすさと互換性のため、英字を推奨する
- `if`、`for`、`class` などの予約語は変数として使用できない

````python
a = 3
b = 5
c = a + b * 7
c
````

##基本的なデータ型

| 型 | 内容 | 例 |
|---|---|---|
| `int` | 整数 | `10` |
| `float` | 小数 | `3.14` |
| `str` | 文字列 | `'山口県'` |
| `bool` | 真偽値 | `True`, `False` |

````python
count = 10
rate = 0.8
city = '周南市'
is_student = True

print(type(count))
print(type(rate))
print(type(city))
print(type(is_student))
````

##算術演算

| 演算子 | 意味 | 例 |
|---|---|---|
| `+` | 加算 | `5 + 2` |
| `-` | 減算 | `5 - 2` |
| `*` | 乗算 | `5 * 2` |

##キーボードからの入力

`input()` の戻り値は、数字を入力した場合でも文字列 `str` である。計算に使うときは、`int()` や `float()` で変換する。

````python
text = input('整数を入力してください: ')
number = int(text)

print('入力した数は', number)
print('2倍すると', number * 2)
````

### 入力エラーへの対処

数字以外を `int()` で変換すると `ValueError` が発生する。次の例では、`try` と `except` を使ってエラーを処理する。

````python
text = input('整数を入力してください: ')

try:
    number = int(text)
    print('2倍すると', number * 2)
except ValueError:
    print('整数として読み取れませんでした。半角数字で入力してください。')
````

##条件分岐：`if`

条件が成立した場合と、成立しなかった場合で処理を分ける。Python ではインデント（字下げ）がプログラムの構造を表す。原則として半角スペース4個を使用する。

````python
n = 20

if n > 10:
    print(n, 'は10より大きいです。')
else:
    print(n, 'は10以下です。')
````

### 比較演算子と論理演算子

| 演算子 | 意味 |
|---|---|
| `==` | 等しい |
| `!=` | 等しくない |
| `<`, `<=`, `>`, `>=` | 大小比較 |
| `and` | 両方の条件が成立 |
| `or` | どちらかの条件が成立 |
| `not` | 条件を反転 |

````python
n = 15

if n > 10 and n < 20:
    print(n, 'は10より大きく、20より小さいです。')
````

より簡潔に、次のようにも書ける。

````python
n = 15

if 10 < n < 20:
    print(n, 'は10より大きく、20より小さいです。')
````

##リスト、タプル、辞書

| データ構造 | 特徴 | 記号 |
|---|---|---|
| リスト `list` | 順序があり、要素を追加・変更できる | `[ ]` |
| タプル `tuple` | 順序があり、作成後に要素を変更できない | `( )` |

````python
stations = ['東京', '品川', '新横浜', '名古屋', '京都', '新大阪']
numbers = [5, 6, 7, 8, 9, 10]

point = (35.0, 135.0)

prices = {
    'りんご': 100,
    'みかん': 300
}

print(stations)
print(point)
print(prices['りんご'])
````

##要素の指定：インデックス

Python では先頭の要素を **0番目**として数える。負の番号は末尾から数える。

- `stations[0]`：先頭
- `stations[2]`：3番目
- `stations[-1]`：最後

````python
print('先頭:', stations[0])
print('3番目:', stations[2])
print('最後:', stations[-1])
````

##スライス：範囲の指定

基本形は次のとおりである。

```python
データ[開始位置:終了位置:ステップ]
```

終了位置の要素は含まれない。`stations[2:4]` は、インデックス2と3、つまり3番目と4番目の要素を取り出す。

````python
stations = ['東京', '品川', '新横浜', '名古屋', '京都', '新大阪', '新神戸', '岡山', '広島', '小倉', '博多']

print(stations[0:5])   # 先頭から5個
print(stations[2:4])   # インデックス2と3
print(stations[3:])    # インデックス3から最後
print(stations[-2:])   # 最後の2個
print(stations[0:8:2]) # 2個おき
````

## `range()` で連続する整数を作る

`range(開始, 終了, ステップ)` は、繰り返し処理でよく使用する。終了の値は含まれない。

````python
print(list(range(3)))        # 0, 1, 2
print(list(range(2, 6)))     # 2, 3, 4, 5
print(list(range(0, 10, 2))) # 0, 2, 4, 6, 8
print(list(range(3, 3)))     # 空
````

##関数

関数は、処理に名前を付けて再利用する仕組みである。`def` で定義し、`return` で結果を返す。

#実行環境（ランタイム）の管理

##処理を中断する

処理が終わらない場合は、実行中セルの左側に表示される停止ボタンを押す。メニューの
**「ランタイム」→「実行を中断」**
から停止できる場合もある。

次の例は、`終了` と入力すると安全に終了する。

````python
while True:
    text = input('文字を入力してください（終了で停止）: ')
    if text == '終了':
        print('処理を終了します。')
        break
    print('入力:', text)
````

##ランタイムの再起動

変数やライブラリの状態を初期化したい場合は、
**「ランタイム」→「セッションを再起動」**
を選ぶ。

##GPU・TPU の選択

機械学習などで必要な場合は、
**「ランタイム」→「ランタイムのタイプを変更」**
からハードウェアアクセラレータを選択する。

- 通常の Python 学習や小規模な表計算では CPU で十分である。
- GPU や TPUのタイプや利用可能時間は変更されることがある。
- GPU を選択しても高速にならないことがある。
- 不要な GPU セッションは終了し、計算資源を節約する。

##セッションと一時ファイル

- `/content` は実行中の仮想マシン上の一時領域である。
- ランタイムが終了すると、変数、アップロードしたファイル、作成した一時ファイルは失われる。
- 残したいファイルは Google Drive にコピーするか、パソコンへダウンロードする。
- Colab のセッション管理画面から不要なセッションを終了できる。

#データファイルの読み込み

##読み込み方法の選択

| 方法 | 向いている場面 | 注意点 |
|---|---|---|
| パソコンから一時アップロード | 小さな演習ファイルをすぐ使う | セッション終了で消える |

##サンプルデータを作る

次のセルは、商品と金額の表を pandas の `DataFrame` として作成する。

````python
import pandas as pd

sample_data = pd.DataFrame({
    '商品': ['りんご', 'みかん', 'すいか'],
    '金額': [300, 200, 1000],
    '個数': [2, 5, 1]
})

sample_data
````

## CSV・Excel ファイルとして保存する

ここでは Colab の一時領域 `/content` に保存する方法を説明する。

````python
from pathlib import Path

work_dir = Path('/content')
csv_path = work_dir / '商品金額.csv' # / はパスを結合する演算子である
excel_path = work_dir / '商品金額.xlsx'

sample_data.to_csv(csv_path, index=False, encoding='utf-8-sig')
sample_data.to_excel(excel_path, index=False)

print(csv_path)
print(excel_path)
````

Colab画面の左端のフォルダーのアイコンをクリックすると、ファイルメニューが開く。
`商品金額.csv` と  
`商品金額.xlsx` が保存されていることを確認せよ。

`pd.read_csv()` を使用する。`shape` は `(行数, 列数)`のタプル、
`size` は全要素数を返す。

````python
print(f'{csv_path} から読み込む')
data_csv = pd.read_csv(csv_path)

print('行数・列数:', data_csv.shape)
print('全要素数:', data_csv.size)
data_csv.head()
````

##Excel ファイルを読み込む

`pd.read_excel()` を使用する。シート名を指定することもできる。

````python
data_excel = pd.read_excel(excel_path)

print('行数・列数:', data_excel.shape)
data_excel.head()
````

##Drive 上のファイルパスを指定する

左端のフォルダマークをクリックし、ファイル一覧を表示する。
対象ファイルのメニュー（右クリックすると表示される）を開き、パスをコピーできる。コピーしたパスを `Path` に設定する。

```python
from pathlib import Path
import pandas as pd

file_path = Path('/content/drive/MyDrive/講義データ/商品金額.xlsx')

if file_path.exists():
    data = pd.read_excel(file_path)
    display(data.head())
else:
    print('ファイルが見つかりません:', file_path)
```

日本語の濁点などが異なる文字として保存されている場合があるため、ファイル名を手入力するよりも「パスをコピー」を使う方が確実である。

##ファイルのダウンロード

作成したファイルを手元のPCへ保存する例である。

```python
from google.colab import files
files.download('/content/商品金額.csv')
```

Google Drive に保存する場合は、Drive をマウントした後でコピーする。

```python
import shutil
shutil.copy('/content/商品金額.csv',
            '/content/drive/MyDrive/講義データ/商品金額.csv')
```
