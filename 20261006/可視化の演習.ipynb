## Medical Cost Personal Datasets の説明

米国の医療保険に関する個人単位のデータである。  
**1行が1人**を表し、年齢、性別、BMI、扶養家族数、喫煙の有無、居住地域、医療費が記録されている。

| 変数名 | 内容 | 種類 |
|---|---|---|
| `age` | 主たる被保険者の年齢 | 数値 |
| `sex` | 性別（female / male） | カテゴリ |
| `bmi` | BMI（Body Mass Index） | 数値 |
| `children` | 健康保険でカバーされている子ども・扶養家族の人数 | 数値・離散 |
| `smoker` | 喫煙の有無（yes / no） | カテゴリ |
| `region` | 米国内の居住地域（northeast / northwest / southeast / southwest） | カテゴリ |
| `charges` | 健康保険に請求された個人の医療費 | 数値 |
