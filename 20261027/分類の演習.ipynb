#分類の演習


##データの概要

本演習では **Bank Marketing Data Set** の `bank-additional-full.csv` を使用する。以下のKaggleのサイトより各自でファイルをダウンロードする。

- Kaggle: `tunguz/bank-marketing-data-set`
- Kaggle URL: https://www.kaggle.com/datasets/tunguz/bank-marketing-data-set

`bank-additional-full.csv` は、2008年5月〜2010年11月の電話マーケティングに関する実績のデータで
**41,188件のレコードと20個の説明変数**
である。

目的変数 `y` は、
- `yes`：定期預金に加入した
- `no`：加入しなかった

である。

##分析のテーマ

分析のテーマはキャンペーンでの架電の効率化である。
まずは、基礎集計を行うことでデータの全体像を理解した上で、
適切にモデルを構築し、架電効率化策を作成せよ。

###注意点

- `unknown` について

  いくつかのカテゴリ変数には文字列 `"unknown"` が含まれる。
  これは `NaN` とは異なる。
  **「値が不明である」というカテゴリ** を意味する。

- `pdays=999` について

  `pdays=999` は「999日前」という通常の意味ではなく、
  **過去のキャンペーンで接触していない**ことを表す特殊値である。
